```python
# ==========================================================
# MATRIZ INVERSA POR EL MÉTODO DE GAUSS-JORDAN
# (Sin utilizar NumPy ni SciPy)
# ==========================================================

def imprimir_matriz(M):
    for fila in M:
        print(["{:.6f}".format(x) for x in fila])
    print()


def matriz_inversa(A):

    n = len(A)

    # Crear matriz identidad
    I = []
    for i in range(n):
        fila = []
        for j in range(n):
            if i == j:
                fila.append(1.0)
            else:
                fila.append(0.0)
        I.append(fila)

    # Matriz aumentada [A|I]
    AI = []

    for i in range(n):
        AI.append(A[i][:] + I[i])

    # Método de Gauss-Jordan
    for i in range(n):

        # Buscar pivote
        pivote = AI[i][i]

        if pivote == 0:

            intercambio = False

            for k in range(i + 1, n):
                if AI[k][i] != 0:
                    AI[i], AI[k] = AI[k], AI[i]
                    intercambio = True
                    break

            if not intercambio:
                print("La matriz no tiene inversa")
                return None

            pivote = AI[i][i]

        # Hacer pivote igual a 1
        for j in range(2 * n):
            AI[i][j] = AI[i][j] / pivote

        # Hacer ceros arriba y abajo del pivote
        for k in range(n):

            if k != i:

                factor = AI[k][i]

                for j in range(2 * n):
                    AI[k][j] = AI[k][j] - factor * AI[i][j]

    # Extraer inversa
    inversa = []

    for i in range(n):
        inversa.append(AI[i][n:])

    return inversa

    A = [
    [2,3],
    [1,4]
]

inv = matriz_inversa(A)

print("Matriz inversa:")

imprimir_matriz(inv)

A = [
    [1,2,3,4],
    [2,1,2,1],
    [3,2,1,2],
    [1,1,1,1]
]

inv = matriz_inversa(A)

print("Matriz inversa:")

imprimir_matriz(inv)
```

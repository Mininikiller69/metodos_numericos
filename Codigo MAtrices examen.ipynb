```python
# =====================================================
# MATRIZ INVERSA MEDIANTE GAUSS-JORDAN
# Sin utilizar NumPy
# =====================================================

def imprimir(M):
    for fila in M:
        print(fila)
    print()


def inversa(A):

    n = len(A)

    # Matriz identidad
    I = []

    for i in range(n):
        fila = []

        for j in range(n):

            if i == j:
                fila.append(1.0)
            else:
                fila.append(0.0)

        I.append(fila)

    # Matriz aumentada
    AI = []

    for i in range(n):
        AI.append(A[i][:] + I[i])

    # Método de Gauss-Jordan
    for i in range(n):

        # Buscar pivote
        if AI[i][i] == 0:

            for k in range(i+1,n):

                if AI[k][i] != 0:

                    AI[i],AI[k]=AI[k],AI[i]

                    break

        pivote = AI[i][i]

        if pivote == 0:
            print("La matriz no tiene inversa.")
            return

        # Normalizar fila
        for j in range(2*n):
            AI[i][j] = AI[i][j]/pivote

        # Eliminar columna
        for k in range(n):

            if k != i:

                factor = AI[k][i]

                for j in range(2*n):

                    AI[k][j] = AI[k][j] - factor*AI[i][j]

    INV = []

    for i in range(n):

        fila=[]

        for j in range(n,2*n):

            valor = AI[i][j]

            if abs(valor-round(valor))<1e-10:
                valor=int(round(valor))

            fila.append(valor)

        INV.append(fila)

    return INV


# ============================================
# EJERCICIO 1
# ============================================

A = [
    [2,-3],
    [-1,1]
]

print("Ejercicio 1")
R = inversa(A)
imprimir(R)

# ============================================
# EJERCICIO 2
# ============================================

A = [
    [4,0,0,5],
    [1,0,4,0],
    [3,4,1,3],
    [1,3,3,0]
]

print("Ejercicio 2")
R = inversa(A)
imprimir(R)
```
